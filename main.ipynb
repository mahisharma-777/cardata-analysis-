```python
#Top 10 most costlier cars in data sets
def format_values(value):
    if value >= 10000000:
        return f"{value/10000000:.2f} Crores"
    elif value >= 100000:
        return f"{value/100000:.2f} Lakhs"
    elif value >= 1000:
        return f"{value/1000:.2f} k"
    else:
        
        return str(int(value))
```

```python
plt.figure(figsize=(13,5))
plt.suptitle("Top 10 most costlier cars", fontsize=20)

top_10_costlier_cars= df.groupby("car_name")['selling_price'].max().sort_values(ascending=False).head(10)

ax = sns.barplot(x=top_10_costlier_cars.index, y=top_10_costlier_cars.values, palette='Set1', ec='black', linewidth=1.5)

formatted_labels = [format_values(i) for i in top_10_costlier_cars.values]
for bar, label in zip(ax.containers, formatted_labels):
    plt.bar_label(container=bar, labels=[label],label_type='edge')

plt.xlabel("Luxury Car Name")
plt.ylabel("Selling Price")
plt.xticks(rotation=90)
```
